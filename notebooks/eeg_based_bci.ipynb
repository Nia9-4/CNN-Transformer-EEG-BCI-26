### Motivation

### Introduction
Brain-Computer-Interfaces (BCI) ... (definition BCI, motivation aka. usage in neurological diseases, EEG)

Limitations of some BCIs 

For practice purposes, the relatively old but publically available dataset 2a from the BCI Competition IV has been used. 

Encoder-only models are transformers that are designed to analyze static input sequences simultaneously. While excelling at classification tasks, given the sequential nature of EEG signals, they are unsuited for decoding movement imagination (MI), and instead, sequence models should be used. Partially inspired by Liao et al. (2025), this project attempts to build a transformer which manages to classify imagined movements with a high accuracy from sequential EEG data. The EEG data was acquired from participants who executed one out of four imaginary movements and was provided as a publically available dataset of the BCI Competition IV (dataset 2a).

3. Third, verify your pipeline by overfitting a single mini-batch. Before training your network
on an entire dataset, select exactly one mini-batch of data and train the model on it for
many epochs. Because the model has more than enough capacity to memorize a single batch,
the training loss should rapidly drop to near zero. If the loss fails to decrease, your codebase
contains a fundamental bug, such as broken gradient flow, mismatched target labels, or
incorrect parameter updates. This diagnostic test is an effective debugging technique in deep
learning development

4. Finally, prioritize rapid experimental iteration. Design your data loading pipelines and vali-

learning is an empirical science that relies on testing hypotheses, adjusting hyperparameters,
and evaluating performance. Building a fast, modular pipeline allows you to systematically
discover optimal hyperparameters and identify architectural bottlenecks far more efficiently
than running single, long-term training runs.
